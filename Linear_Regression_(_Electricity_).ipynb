```python
df = pd.read_csv('electricity.csv')
df.tail()
```

```python
df.shape
```

```python
df_subset = df.head(4)
plt.bar(df_subset['ac_hours_per_day'].astype(str), df_subset['monthly_electricity_bill'])
plt.xlabel('AC Hours per Day')
plt.ylabel('Monthly Electricity Bill')
plt.title('Monthly Electricity Bill for First 4 Rows')
plt.show()
```

```python
x = df.drop('monthly_electricity_bill', axis=1)
y = df['monthly_electricity_bill']
y.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
le = LinearRegression()
le.fit(x_train, y_train)
```

```python
le.predict([[6, 11, 970]]).round(2)
```

```python
print(le.score(x_test, y_test) * 100)
```
